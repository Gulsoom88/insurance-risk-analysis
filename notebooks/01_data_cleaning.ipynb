```python
import sqlite3

conn = sqlite3.connect("../insurance.db")
```

```python
df.to_sql("insurance_policies", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT
    policy_type,
    AVG(premium_amount) AS avg_premium,
    AVG(claim_history) AS avg_claim_history,
    AVG(risk_profile) AS avg_risk
FROM insurance_policies
GROUP BY policy_type
"""

result = pd.read_sql(query, conn)

result
```

```python
# Create age groups
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)

# Create credit score groups
df["credit_score_group"] = pd.cut(
    df["credit_score"],
    bins=[0, 580, 670, 740, 800, 900],
    labels=["Poor", "Fair", "Good", "Very Good", "Excellent"]
)

# Create premium to coverage ratio
df["premium_coverage_ratio"] = df["premium_amount"] / df["coverage_amount"]

# Create estimated customer value
df["estimated_value"] = df["premium_amount"] * (1 + df["risk_profile"])

# Create risk category
df["risk_category"] = pd.cut(
    df["risk_profile"],
    bins=[-1, 0, 1, 2, 3],
    labels=["Very Low", "Low", "Medium", "High"]
)
```

```python
df.to_csv("../data/processed/insurance_clean.csv", index=False)
```
